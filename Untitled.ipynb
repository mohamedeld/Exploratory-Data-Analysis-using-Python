```python
rng = np.random.default_rng(30)
customer_ages = np.random.normal(loc=35, scale=10, size=500)

fig , ax = plt.subplots(figsize=(12,6))
ax.hist(customer_ages, bins=20, color="skyblue", edgecolor="black")

ax.set_title("Districbution for Customer Ages", fontsize=14)
ax.set_xlabel("Age",fontsize=12)
ax.set_ylabel("frequency", fontsize=12)

plt.show()
```

```python
fig , ax = plt.subplots(figsize=(12,6))
ax.boxplot(customer_ages, patch_artist=True, boxprops=dict(facecolor="lightgreen", color="black"))

ax.set_title("boxplot of Customer Ages", fontsize=14)
ax.set_xlabel("Age", fontsize=12)

plt.show()
```
