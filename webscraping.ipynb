```python
from urllib.request import Request,urlopen
```

```python
url = 'https://eg.opensooq.com/ar/%D8%B3%D9%8A%D8%A7%D8%B1%D8%A7%D8%AA-%D9%88%D9%85%D8%B1%D9%83%D8%A8%D8%A7%D8%AA/%D8%B3%D9%8A%D8%A7%D8%B1%D8%A7%D8%AA-%D9%84%D9%84%D8%A8%D9%8A%D8%B9/%D9%85%D8%B1%D8%B3%D9%8A%D8%AF%D8%B3-%D8%A8%D9%86%D8%B2'
```

```python
client = urlopen(url)
```

```python
html = client.read()
```

```python
client.close()
```

```python
soup = bs(html, "html.parser")
```

```python
containers = soup.find_all("a",{"class":"sc-a0c844ea-0"})
```

```python
len(containers)
```

```python
import re
```

```python
columns = [
    "carname",
    "Year",
    "Make",
    "Class",
    "Trim",
    "Condition",
    "Mileage",
    "Fuel Type",
    "Specs",
    "location",
    "price"
]

data = []

for container in containers:
    h2 = container.find("h2")
    spans = container.select(".starCpsItem span")
    location = container.select_one(".locationText")
    price = container.select_one(".redColor")
    if h2:
        row = {
            "carname": h2.get_text(" ", strip=True),
            "location": location.get_text(" ", strip=True) if location else None,
            "price": price.get_text(" ", strip=True) if price else None
        }

        for column, span in zip(columns[1:], spans):
            row[column] = span.get_text(" ", strip=True)

        data.append(row)

df = pd.DataFrame(data, columns=columns)

df
```
