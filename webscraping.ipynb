```python
from bs4 import BeautifulSoup as bs 
```

```python
from urllib.request import Request,urlopen
```

```python
url = 'https://eg.opensooq.com/ar/%D8%B3%D9%8A%D8%A7%D8%B1%D8%A7%D8%AA-%D9%88%D9%85%D8%B1%D9%83%D8%A8%D8%A7%D8%AA/%D8%B3%D9%8A%D8%A7%D8%B1%D8%A7%D8%AA-%D9%84%D9%84%D8%A8%D9%8A%D8%B9/%D9%85%D8%B1%D8%B3%D9%8A%D8%AF%D8%B3-%D8%A8%D9%86%D8%B2'
```

```python
client = urlopen(url)
```

```python
html = client.read()
```

```python
client.close()
```

```python
soup = bs(html, "html.parser")
```

```python
containers = soup.find_all("a",{"class":"sc-a0c844ea-0"})
```

```python
len(containers)
```

```python
data = []
for container in containers:
    h2 = container.find("h2")
    if h2:
        data.append({
            "carname":h2.get_text(strip=True)
        })
df = pd.DataFrame(data)
df
```
