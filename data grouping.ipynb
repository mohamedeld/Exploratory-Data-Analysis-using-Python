```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic.keys()
```

```python
titanic.groupby("class").groups
```

```python
titanic.groupby("class").get_group('First')
```

```python
titanic.groupby("class")['survived'].count()
```

```python
titanic.groupby("sex")['age'].count()
```

```python
titanic.groupby(['sex','class']).get_group(('female','Third'))
```

```python
titanic.groupby(['sex','class']).sum(numeric_only=True)
```

```python
titanic['mean_age'] = titanic.groupby(['sex','pclass'])['age'].transform('mean')
titanic['mean_age']
```

```python
titanic.age.fillna(titanic.mean_age)
```
