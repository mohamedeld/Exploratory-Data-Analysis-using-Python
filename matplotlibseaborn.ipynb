```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]

temp_cairo = [30, 31, 29, 32, 30, 28, 31]

temp_luxor = [36, 38, 37, 39, 36, 35, 38]
```

```python
df_cairo = pd.DataFrame(data={'Days':days, 'Temperatures':temp_cairo})
```

```python
df_luxor = pd.DataFrame(data={'Days':days, 'Temperatures':temp_luxor})
```

```python
df_cairo
```

```python
plt.plot(df_cairo.Days,df_cairo.Temperatures,data=df_cairo,label="Cairo",ls="--",lw=2, alpha=0.5)
plt.plot(df_luxor.Days,df_luxor.Temperatures,label="Luxor", c="g", marker="s", markersize=10)

plt.title("Tems in Cairo & Luxor", fontsize=20)
plt.xlabel("Days")
plt.ylabel("Temps")
# plt.grid()
# plt.ylim(0,100)
plt.legend()
plt.savefig("data/graph.png")
```

```python
plt.scatter(df_cairo.Days,df_cairo.Temperatures)
plt.scatter(df_luxor.Days,df_cairo.df_luxor)

plt.title("Tems in Cairo & Luxor", fontsize=20)
```

```python
penguins = sns.load_dataset("penguins")
```

```python
penguins.head()
```

```python
penguins.info()
```

```python
penguins.columns
```

```python
penguins_df = penguins.dropna(
    subset=['species', 'island', 'bill_length_mm', 'bill_depth_mm',
       'flipper_length_mm', 'body_mass_g', 'sex']
)
penguins_df.info()
```

```python
penguins_df.isna().sum()
```

```python
sns.scatterplot(
    data=penguins_df,
    x="bill_length_mm",
    y="flipper_length_mm"
)
plt.title("Seaborn scatter plot")
plt.show()
```

```python
sns.scatterplot(
    data=penguins_df,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species"
)
plt.title("Seaborn scatter plot")
plt.show()
```

```python
fig , ax = plt.subplots(figsize=(7,5))

sns.scatterplot(
    data=penguins_df,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    ax=ax
)
ax.set_title("Seaborn scatter plot")
ax.set_xlabel("Bill Length (mm)")
ax.set_ylabel("Flipper Length (mm)")
plt.show()
```

```python
g = sns.relplot(
    data=penguins_df,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    height=5,
    aspect=1.3,
    col="island",
    kind="scatter"
)
g.fig.suptitle("Figure-Level Function: sns.relplot()",y=1.03)
plt.show()
```
