```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
cars.rename(columns={
   'model year':'model_year'
},inplace=True)
```

```python
cars.origin.value_counts()
```

```python
cars['origin'] = cars['origin'].replace({
    'United States': 'usa'
})
```

```python
cars.origin.value_counts()
```

```python
cars.horsepower = cars.horsepower.str.split(' ').str.get(0)
```

```python
cars.horsepower.unique()
```

```python
cars.horsepower = pd.to_numeric(
    cars.horsepower,
    errors="coerce"
)
```

```python
cars.describe()
```

```python
cars.name = cars.name.str.lower().str.strip()
```

```python
cars.model_year.value_counts()
```

```python
cars.model_year = cars.model_year.replace({
    1973:73
})
```

```python
cars.model_year.value_counts()
```

```python
cars.weight.min()
```

```python
cars.loc[cars.weight.idxmax(),'weight'] = 2300
```

```python
cars.weight.describe()
```

```python
cars.mpg.min()
```

```python
cars.loc[cars.mpg.idxmin(),'mpg'] = 1 / cars.loc[cars.mpg.idxmin(),'mpg']
```
